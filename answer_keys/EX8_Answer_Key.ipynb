````python
from importlib.util import module_from_spec, spec_from_file_location
from pathlib import Path
from tempfile import TemporaryDirectory

module_source = '''
class Observation:
    def __init__(self, high, low, dew, pres, unit="F"):
        if unit not in ("F", "C", "K"):
            raise ValueError("unit must be F, C, or K")
        self.high_temperature = high
        self.low_temperature = low
        self.dewpoint = dew
        self.pressure = pres
        self.unit = unit

    def F_to_K(self):
        return (self.high_temperature + 459.67) * (5 / 9)

    def F_to_C(self):
        return (self.high_temperature - 32) * (5 / 9)

    def C_to_F(self):
        return self.high_temperature * 9 / 5 + 32

    def C_to_K(self):
        return self.high_temperature + 273.15

    def K_to_F(self):
        return (self.high_temperature - 273.15) * 9 / 5 + 32

    def K_to_C(self):
        return self.high_temperature - 273.15

    def formatted_conversion(self, conversion):
        options = {
            "F": {"F_to_C": self.F_to_C, "F_to_K": self.F_to_K},
            "C": {"C_to_F": self.C_to_F, "C_to_K": self.C_to_K},
            "K": {"K_to_F": self.K_to_F, "K_to_C": self.K_to_C},
        }[self.unit]
        if conversion not in options:
            first, second = options
            return f"incorrect unit. Expected '{first}' or '{second}'"

        return f"{options[conversion]():.2f}"
'''

with TemporaryDirectory() as directory:
    module_path = Path(directory) / "my_file.py"
    module_path.write_text(module_source, encoding="utf-8")
    spec = spec_from_file_location("my_file", module_path)
    module = module_from_spec(spec)
    spec.loader.exec_module(module)
    Observation = module.Observation

obs = Observation(high=85, low=65, dew=64, pres=29.4)

print("The high temperature is", obs.high_temperature)
print("The low temperature is", obs.low_temperature)
print("The dewpoint is", obs.dewpoint)
print("The pressure is", obs.pressure)
print("High in K =", obs.F_to_K())
print("High in C =", obs.F_to_C())
````
