### Problem 7

Create a function named ```p7``` that follows these requirements:

1. It takes one parameter named ```CAPE```.
2. For all cases, print ```"CAPE is XXXX j kg-1"```, where XXXX is the value of the ```CAPE``` parameter. For example, if the CAPE parameter is 50, print ```"CAPE is 50 j kg-1"```. Display `int(CAPE)` in the message and use the original value for the category comparisons.
3. If ```CAPE``` is greater than or equal to 2000, add ```(Extreme Instability).``` to the end of the string.
4. If ```CAPE``` is greater than or equal to 1000 but less than 2000, add ```(Moderate Instability).``` to the end of the string.
5. If ```CAPE``` is greater than 0, but less than 1000, add ```(Low Instability).``` to the end of the string.
6. If ```CAPE``` is 0, add ```(No Instability).``` to the end of the string.
7. If ```CAPE``` is less than 0, add ```(Invalid. Check Input).``` to the end of the string.

Example output:

```
CAPE is 2000 j kg-1 (Extreme Instability).
CAPE is 1000 j kg-1 (Moderate Instability).
CAPE is 0 j kg-1 (Low Instability).
CAPE is 0 j kg-1 (No Instability).
CAPE is -10 j kg-1 (Invalid. Check Input).
```

````python
def p7(CAPE):
    if CAPE >= 2000:
        category = "Extreme Instability"
    elif CAPE >= 1000:
        category = "Moderate Instability"
    elif CAPE > 0:
        category = "Low Instability"
    elif CAPE == 0:
        category = "No Instability"
    else:
        category = "Invalid. Check Input"

    print(f"CAPE is {int(CAPE)} j kg-1 ({category}).")


for value in (2000, 1000, 0.01, 0, -10):
    p7(value)
````
