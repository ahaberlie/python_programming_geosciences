## Task 1

Convert the following python ```lists``` into ```numpy.ndarrays```.

```dkb_high_temps``` are average high temps for each month in DeKalb in Fahrenheit

```dkb_low_temps``` are average low temps for each month in DeKalb in Fahrenheit

```dkb_precip``` are average precipitation totals for each month in DeKalb in inches

````python
import numpy as np

dkb_high_temps = [28, 32, 45, 59, 70, 80, 83, 81, 75, 62, 46, 34]

dkb_low_temps = [13, 17, 27, 37, 49, 60, 63, 61, 53, 41, 30, 19]

dkb_precip = [1.69, 1.71, 2.24, 3.56, 4.79, 4.57, 4.00, 4.06, 3.32, 2.96, 2.31, 2.04]

# replace None with your code
dkb_high_npy = np.array(dkb_high_temps)

# replace None with your code
dkb_low_npy = np.array(dkb_low_temps)

# replace None with your code
dkb_precip_npy = np.array(dkb_precip)
````

## Task 2

Convert the monthly rainfall totals in `dkb_precip_npy` from inches to millimeters. Use 25.4 mm per inch.

````python
dkb_precip_mm = np.multiply(dkb_precip_npy, 25.4)
dkb_precip_mm
````

## Task 3

Estimate the mean daily temperature for each month as the midpoint of the two monthly average temperatures.

````python
dkb_average = np.mean([dkb_high_npy, dkb_low_npy], axis=0)
dkb_average
````

## Task 4

Find the difference between high and low temperatures for each month using ```dkb_high_npy``` and ```dkb_low_npy```.

````python
dkb_difference = np.subtract(dkb_high_npy, dkb_low_npy)
dkb_difference
````
