# Task 1 - Implement "helper" functions

I will provide you with the following "abstract functions" (incomplete functions that you need to finish) that you will need to modify and complete.

````python
#example: my example function that finds the count of values above 32.
def hw3_count_above_freezing(temperatures):  
    '''
    Return the count of how often a value in the list that is passed to the function
    is greater than 32.

    The for loop used in this list **must** use an iteration variable named ```temperature```.

    The "running count" variable **must** be named ```count```.

            Parameters:
                    temperatures (list): A list of temperatures that are either ints or floats

            Returns:
                    count (int): A count of how many times there was a value greater than 
                    32 in ```temperatures```
    '''
    count = 0

    for temperature in temperatures:

        if temperature > 32:
            count += 1

    return count

def hw3_mean(values):
    """Return the mean of a nonempty list."""
    value_sum = 0
    value_count = 0
    for value in values:
        value_sum += value
        value_count += 1
    value_mean = value_sum / value_count
    return value_mean


def hw3_frequency(values, threshold):
    """Count values strictly above the threshold."""
    exceedances = 0
    for value in values:
        if value > threshold:
            exceedances += 1
    return exceedances


def hw3_minimum(values):
    """Return the minimum value from a nonempty list."""
    minimum_value = float("inf")
    for value in values:
        if value < minimum_value:
            minimum_value = value
    return minimum_value


def hw3_maximum(values):
    """Return the maximum value from a nonempty list."""
    maximum_value = float("-inf")
    for value in values:
        if value > maximum_value:
            maximum_value = value
    return maximum_value
````
