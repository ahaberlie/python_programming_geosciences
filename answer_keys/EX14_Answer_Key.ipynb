### Problem 1

Add a title to the following plot with the text "Problem 1"

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(1, 1, "k*")
ax.set_title("Problem 1")
plt.show()
````

### Problem 2

Add a label to the y axis with the text "unit y"

Add a label to the x axis with the text "unit x"

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(1, 1, "r*")
ax.set(xlabel="unit x", ylabel="unit y")
plt.show()
````

### Problem 8

Draw a triangle using the ```plot``` method.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
x = [1, 3, 2, 1]
y = [1, 1, 3, 1]
ax.plot(x, y)
ax.set_aspect("equal")
plt.show()
````
