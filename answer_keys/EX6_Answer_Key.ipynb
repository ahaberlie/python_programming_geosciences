### Problem 6

Use `enumerate` to get each zero-based index and value. Use `if` / `else` to print the index when the value is 5 and skip every other value.

````python
a = [1, 2, 3, 4, 5]

for index, value in enumerate(a):
    if value == 5:
        print(index)
    else:
        continue
````

## ```dict``` iterators

In the following problems, you must use **one** of these dictionary iterators: ```keys```, ```values```, or ```items```. 

You must include the message building code within the loop, and it should work for any sized ```dict```.

### Problem 7

Use **one iterator loop** using the defined ```dict``` named ```obs``` to print out the following messages:

```
The temperature on Jan 1 was 15 F
The temperature on Jan 2 was 16 F
The temperature on Jan 3 was 17 F
```

````python
obs = {'Jan 1': 15, 'Jan 2': 16, 'Jan 3': 17}

for date, temperature in obs.items():
    print(f'The temperature on {date} was {temperature} F')
````

### Problem 8

Use **one iterator loop** using the defined ```dict``` named ```obs``` to print out the following conditional messages using an ```if / else``` statement:

```
The temperature on Jan 1 was 45 F
The temperature on Jan 2 was 32 F. It was freezing!
The temperature on Jan 3 was 15 F. It was freezing!
```

````python
obs = {"Jan 1": 45, "Jan 2": 32, "Jan 3": 15}

for date, temperature in obs.items():
    message = f"The temperature on {date} was {temperature} F"
    if temperature <= 32:
        print(message + ". It was freezing!")
    else:
        print(message)
````
