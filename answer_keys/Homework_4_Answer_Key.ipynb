### Task 2

Code that creates a graph like your line graph example:

````python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
daily_highs_f = [68, 72, 76, 80, 74, 70, 73]

fig, ax = plt.subplots()
ax.plot(days, daily_highs_f, marker="o")
ax.set(title="Daily high temperatures · illustrative station",
       xlabel="Day of June", ylabel="Temperature (°F)", xticks=days)
plt.show()
````

### Task 3

Code that creates a graph like your bar graph example:

````python
import matplotlib.pyplot as plt

years = ["2022", "2023", "2024", "2025"]
june_totals_mm = [84, 122, 95, 108]

fig, ax = plt.subplots()
ax.bar(years, june_totals_mm)
ax.set(title="June precipitation totals · illustrative station",
       xlabel="Year", ylabel="Precipitation (mm)")
plt.show()
````

### Task 4

Code that creates a graph like your scatterplot graph example:

````python
import matplotlib.pyplot as plt

high_f = [68, 72, 76, 80, 74, 70, 73]
dew_point_f = [48, 53, 55, 61, 58, 52, 54]

fig, ax = plt.subplots()
ax.scatter(high_f, dew_point_f)
ax.set(title="Daily high and dew point · illustrative station",
       xlabel="Daily high (°F)", ylabel="Afternoon dew point (°F)")
plt.show()
````
