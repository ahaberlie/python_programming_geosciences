### Problem 7

turn the list of rainfall in mm into a numpy array

next, Turn the list of zeros and ones into a NumPy array

finally, find the sum of `a_n` *only* where `b_n` is 1

'a_n' must be an ndarray

'b_n' must be an ndarray

'c' must be the result where positions in 'a_n' that correspond to a 0 in 'b_n' are also zero

'd' must be the correct sum

```python
def problem_7():

    a = [[10, 11, 12, 13, 14],
         [11, 12, 13, 14, 15],
         [12, 13, 14, 15, 16],
         [13, 14, 15, 16, 17],
         [14, 15, 16, 17, 18]]

    a_n = np.array(a)

    b = [[1, 0, 0, 0, 1],
         [0, 1, 0, 1, 0],
         [0, 0, 1, 0, 0],
         [0, 1, 0, 1, 0],
         [1, 0, 0, 0, 1]]

    b_n = np.array(b)

    c = a_n * b_n

    c_sum = np.sum(c)

    return a_n, b_n, c, c_sum

print(problem_7())
```
